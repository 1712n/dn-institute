```python
def price_after_hours(trades, buy_time, hours):
    target_time = buy_time + pd.Timedelta(hours=hours)

    future = trades[
        trades["date"] >= target_time
    ].sort_values("date")

    if len(future) == 0:
        return None

    return future.iloc[0]["price"]
```

```python
def price_after_hours(trades, buy_time, hours, outcome):
    target_time = buy_time + pd.Timedelta(hours=hours)

    future = trades[
        (trades["date"] >= target_time) &
        (trades["outcome"] == outcome)
    ].sort_values("date")

    if len(future) == 0:
        return None

    return future.iloc[0]["price"]
```

```python
import requests
import time

top30_trades = []

for i, condition_id in enumerate(top30["conditionId"], 1):
    cursor = None
    market_count = 0

    while True:
        params = {
            "condition": condition_id,
            "limit": 1000
        }
        if cursor:
            params["cursor"] = cursor

        r = requests.get(
            "https://data-api.polymarket.com/v2/trades",
            params=params
        )

        if r.status_code != 200:
            print("Error:", i, r.status_code, condition_id)
            break

        data = r.json()
        trades = data.get("data", [])
        top30_trades.extend(trades)
        market_count += len(trades)

        cursor = data.get("pagination", {}).get("next_cursor")

        if not cursor:
            break

        time.sleep(0.1)

    print(f"{i}/30 | trades: {market_count}")

print("TOTAL TRADES:", len(top30_trades))
```

```python
import ast

def get_winner(outcomes, prices):
    try:
        if isinstance(outcomes, str):
            outcomes = ast.literal_eval(outcomes)
        if isinstance(prices, str):
            prices = ast.literal_eval(prices)

        prices = [float(x) for x in prices]

        if 1.0 in prices:
            return outcomes[prices.index(1.0)]
    except:
        pass

    return None

market_results["winning_outcome"] = market_results.apply(
    lambda x: get_winner(x["outcomes"], x["outcomePrices"]),
    axis=1
)

print(
    market_results[
        ["question", "winning_outcome"]
    ].head(15).to_string(index=False)
)
```

```python
def max_price_next_6h(trades, buy_time, outcome):
    end_time = buy_time + pd.Timedelta(hours=6)

    future = trades[
        (trades["date"] > buy_time) &
        (trades["date"] <= end_time) &
        (trades["outcome"] == outcome)
    ]

    if len(future) == 0:
        return None

    return future["price"].max()


shortlist_trades["max_price_6h"] = shortlist_trades.apply(
    lambda x: max_price_next_6h(
        top30_period[
            top30_period["condition_id"] == x["condition_id"]
        ],
        x["date"],
        x["outcome"]
    ),
    axis=1
)

shortlist_trades["move_6h"] = (
    shortlist_trades["max_price_6h"]
    - shortlist_trades["price"]
)

print(
    shortlist_trades["move_6h"].describe()
)
```

```python
final_scores = (
    shortlist_trades
    .groupby("proxy_wallet")
    .agg(
        markets=("condition_id", "nunique"),
        trades=("proxy_wallet", "size"),
        total_notional=("notional", "sum"),
        avg_entry=("price", "mean"),
        avg_hours_before=("hours_to_resolution", "mean"),
        avg_move_6h=("move_6h", "mean")
    )
    .reset_index()
)

# Базовый score
final_scores["score"] = (
    final_scores["markets"] * 20
    + final_scores["avg_move_6h"] * 1000
    + (final_scores["total_notional"] / 100000) * 5
)

# Штраф за очень большое количество сделок
final_scores.loc[
    final_scores["trades"] > 200,
    "score"
] -= 20

final_ranking = (
    final_scores
    .sort_values("score", ascending=False)
    .head(15)
)

print(
    final_ranking.to_string(index=False)
)
```
